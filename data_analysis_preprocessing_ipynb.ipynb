```python
#3. Write a Python program to:
#·  Load a structured dataset (CSV file)
#·  Load unstructured text data (text file)
#·  Display summary statistics for structured data
#·  Display word frequency for text data
```

```python
import pandas as pd
from collections import Counter
import string
df = pd.read_csv("cardata.csv")
print("Structured Dataset Loaded Successfully\n")
with open("Text file sample.txt", "r", encoding="utf-8") as file:
    text_data = file.read()
print("Unstructured Text Data Loaded Successfully\n")
print("Summary Statistics for Structured Data:")
print(df.describe())
print("\n")
text_data = text_data.lower()
text_data = text_data.translate(str.maketrans("", "", string.punctuation))
words = text_data.split()
word_frequency = Counter(words)
print("Word Frequency in Text Data:")
for word, count in word_frequency.items():
    print(f"{word}: {count}")
```
